```python
%matplotlib inline

import matplotlib.pyplot as plt
import pandas as pd

from code import load_and_clean, plot_fig1, plot_fig2, plot_fig3, plot_fig4, plot_fig5, plot_fig6, plot_fig8, plot_fig12, plot_fig13, plot_fig14

file_path = '/content/drive/My Drive/colab/subsidies/data/subsidies_raw.xlsx'
global_product_df, global_total_df, country_product_df, country_total_df = load_and_clean(file_path)

country_elec_df = country_product_df[country_product_df["Product"] == "Electricity"].copy()
country_fossil_df = country_product_df[country_product_df["Product"] != "Electricity"].copy()
country_fossil_agg_df = country_fossil_df.groupby(["Country", "Year"])["Value (B USD)"].sum().reset_index()
global_elec_df = global_product_df[global_product_df["Product"] == "Electricity"].copy()
global_fossil_df = global_product_df[global_product_df["Product"] != "Electricity"].copy()
global_fossil_agg_df = global_fossil_df.groupby("Year")["Value (B USD)"].sum().reset_index()

plot_fig1(global_elec_df, global_fossil_agg_df)
plot_fig2(global_elec_df, global_fossil_agg_df)
plot_fig3(global_elec_df, global_fossil_df)
plot_fig4(global_product_df)
plot_fig5(global_fossil_agg_df, global_elec_df)
plot_fig6(global_fossil_df)
plot_fig8(global_product_df)
plot_fig12(global_product_df)
plot_fig13(country_product_df)
plot_fig14(country_elec_df, country_fossil_df, country_total_df)
```
